```python
import numpy as np
import matplotlib.pyplot as plt
from pynq import Overlay, allocate
import time
import os
from PIL import Image
import glob

# ====== 檔名與 IP 名稱 =======
BITFILE = "gtsrb_cnn.bit"
IP_NAME = "gtsrb_cnn_0"
WEIGHT_FILES = [
    ("weight/weight_0.dat", "w_c1"), ("weight/weight_1.dat", "b_c1"),
    ("weight/weight_2.dat", "w_c2"), ("weight/weight_3.dat", "b_c2"),
    ("weight/weight_4.dat", "w_c3"), ("weight/weight_5.dat", "b_c3"),
    ("weight/weight_6.dat", "w_fc1"), ("weight/weight_7.dat", "b_fc1"),
    ("weight/weight_8.dat", "w_fc2"), ("weight/weight_9.dat", "b_fc2"),
]

# ====== register map =======
REG = {
    "AP_CTRL": 0x00,
    "GIE": 0x04, "IER": 0x08, "ISR": 0x0C,
    "w_c1": 0x10, "b_c1": 0x1C, "w_c2": 0x28, "b_c2": 0x34,
    "w_c3": 0x40, "b_c3": 0x4C, "w_fc1": 0x58, "b_fc1": 0x64,
    "w_fc2": 0x70, "b_fc2": 0x7C, "feature_in": 0x88,
    "logits_out": 0x94, "top1": 0xA0,
}

# ====== GTSRB 類別（43 類，根據 GTSRB 文檔） =======
gtsrb_labels = [
    "20km/h", "30km/h", "50km/h", "60km/h", "70km/h", "80km/h", "80km/h end", 
    "100km/h", "120km/h", "No overtaking", "No overtaking for trucks", 
    "Priority at next intersection", "Priority road", "Give way", "Stop", 
    "No vehicles", "No entry for vehicles >3.5t", "No entry", "General caution", 
    "Dangerous curve left", "Dangerous curve right", "Double curve", 
    "Bumpy road", "Slippery road", "Road narrows", "Road works", 
    "Traffic signals", "Pedestrians", "Children crossing", "Bicycles crossing", 
    "Beware of ice/snow", "Wild animals crossing", "End of restrictions", 
    "Turn right ahead", "Turn left ahead", "Ahead only", "Go straight or right", 
    "Go straight or left", "Keep right", "Keep left", "Roundabout mandatory", 
    "End of no overtaking", "End of no overtaking for trucks"
]

# ====== 網路結構常數 =======
IN_H, IN_W, IN_C = 30, 30, 3
C1_KH, C1_KW, C1_IN, C1_OUT = 3, 3, 3, 16
C1_OH, C1_OW = IN_H - C1_KH + 1, IN_W - C1_KW + 1  # 28x28
P1_KH, P1_KW, P1_IN = 2, 2, C1_OUT
P1_OH, P1_OW = C1_OH // 2, C1_OW // 2  # 14x14
C2_KH, C2_KW, C2_IN, C2_OUT = 3, 3, P1_IN, 32
C2_OH, C2_OW = P1_OH - C2_KH + 1, P1_OW - C2_KW + 1  # 12x12
P2_KH, P2_KW, P2_IN = 2, 2, C2_OUT
P2_OH, P2_OW = C2_OH // 2, C2_OW // 2  # 6x6
C3_KH, C3_KW, C3_IN, C3_OUT = 3, 3, P2_IN, 64
C3_OH, C3_OW = P2_OH - C3_KH + 1, P2_OW - C3_KW + 1  # 4x4
P3_KH, P3_KW, P3_IN = 2, 2, C3_OUT
P3_OH, P3_OW = C3_OH // 2, C3_OW // 2  # 2x2
FC1_IN, FC1_OUT = P3_OH * P3_OW * P3_IN, 64  # 2*2*64=256
FC2_IN, FC2_OUT = FC1_OUT, 43  # GTSRB has 43 classes

# ====== 圖像預處理 =======
def preprocess_image(image_path, size=30):
    if not os.path.exists(image_path):
        print(f"警告: 圖像 {image_path} 不存在，跳過")
        return None
    img = Image.open(image_path).convert("RGB")
    img = img.resize((size, size), Image.LANCZOS)
    return np.array(img).astype("float32") / 255.0

# ====== 軟體前向推理 =======
def conv_valid_relu_numpy(feature_in, W, B, Hin, Win, Cin, Kh, Kw, Cout):
    Hout = Hin - Kh + 1
    Wout = Win - Kw + 1
    out = np.zeros((Hout, Wout, Cout), dtype=np.float32)
    for cout in range(Cout):
        for h in range(Hout):
            for w in range(Wout):
                acc = B[cout]
                for kh in range(Kh):
                    ih = h + kh
                    for kw in range(Kw):
                        iw = w + kw
                        for cin in range(Cin):
                            acc += feature_in[ih, iw, cin] * W[kh, kw, cin, cout]
                if acc < 0:
                    acc = 0.0
                out[h, w, cout] = acc
    return out

def maxpool2x2_numpy(feature_in):
    Hin, Win, Cin = feature_in.shape
    Hout, Wout = Hin // 2, Win // 2
    out = np.zeros((Hout, Wout, Cin), dtype=np.float32)
    for c in range(Cin):
        for h in range(Hout):
            for w in range(Wout):
                h0, w0 = h * 2, w * 2
                out[h, w, c] = np.max(feature_in[h0:h0+2, w0:w0+2, c])
    return out

def linear_numpy(vin, W, b, relu_en=False):
    out = vin.dot(W) + b
    return np.maximum(out, 0) if relu_en else out

def softmax(x):
    ex = np.exp(x - np.max(x))
    return ex / np.sum(ex)

def run_software_forward(weights, input_image):
    Wc1 = weights['w_c1'].reshape((C1_KH, C1_KW, C1_IN, C1_OUT))
    Bc1 = weights['b_c1'].reshape((C1_OUT,))
    c1 = conv_valid_relu_numpy(input_image, Wc1, Bc1, IN_H, IN_W, IN_C, C1_KH, C1_KW, C1_OUT)
    p1 = maxpool2x2_numpy(c1)

    Wc2 = weights['w_c2'].reshape((C2_KH, C2_KW, C2_IN, C2_OUT))
    Bc2 = weights['b_c2'].reshape((C2_OUT,))
    c2 = conv_valid_relu_numpy(p1, Wc2, Bc2, P1_OH, P1_OW, C2_IN, C2_KH, C2_KW, C2_OUT)
    p2 = maxpool2x2_numpy(c2)

    Wc3 = weights['w_c3'].reshape((C3_KH, C3_KW, C3_IN, C3_OUT))
    Bc3 = weights['b_c3'].reshape((C3_OUT,))
    c3 = conv_valid_relu_numpy(p2, Wc3, Bc3, P2_OH, P2_OW, C3_IN, C3_KH, C3_KW, C3_OUT)
    p3 = maxpool2x2_numpy(c3)

    flat = p3.reshape(-1)
    Wf1 = weights['w_fc1'].reshape((FC1_IN, FC1_OUT))
    Bf1 = weights['b_fc1'].reshape((FC1_OUT,))
    fc1 = linear_numpy(flat, Wf1, Bf1, relu_en=True)

    Wf2 = weights['w_fc2'].reshape((FC2_IN, FC2_OUT))
    Bf2 = weights['b_fc2'].reshape((FC2_OUT,))
    logits = linear_numpy(fc1, Wf2, Bf2, relu_en=False)
    return logits, int(np.argmax(logits))

# ====== 載入權重 ======
def load_weights_from_dat():
    w = {}
    for fn, key in WEIGHT_FILES:
        if not os.path.exists(fn):
            raise FileNotFoundError(f"缺少權重檔 {fn}")
        w[key] = np.loadtxt(fn, dtype=np.float32)
    return w

# ====== 64-bit 位址寫入幫手 ======
def write_addr(ip, base, phys_addr):
    low = np.uint32(phys_addr & 0xFFFFFFFF)
    high = np.uint32((phys_addr >> 32) & 0xFFFFFFFF)
    ip.write(base + 0, int(low))
    ip.write(base + 4, int(high))

# ====== 硬體呼叫 ======
def run_hardware_forward(overlay, weights, input_image):
    ip = getattr(overlay, IP_NAME)
    bufs = {}
    for k in ['w_c1', 'b_c1', 'w_c2', 'b_c2', 'w_c3', 'b_c3', 'w_fc1', 'b_fc1', 'w_fc2', 'b_fc2']:
        bufs[k] = allocate(shape=weights[k].shape, dtype=np.float32)
        bufs[k][:] = weights[k]

    bufs['feature_in'] = allocate(shape=(IN_H * IN_W * IN_C,), dtype=np.float32)
    bufs['logits_out'] = allocate(shape=(FC2_OUT,), dtype=np.float32)
    bufs['feature_in'][:] = input_image.reshape(-1)

    for key in ['w_c1', 'b_c1', 'w_c2', 'b_c2', 'w_c3', 'b_c3', 'w_fc1', 'b_fc1', 'w_fc2', 'b_fc2',
                'feature_in', 'logits_out']:
        write_addr(ip, REG[key], bufs[key].physical_address)

    t0 = time.time()
    ap = ip.read(REG["AP_CTRL"])
    ip.write(REG["AP_CTRL"], (ap & 0x80) | 0x01)
    while (ip.read(REG["AP_CTRL"]) & 0x2) == 0:
        pass
    t1 = time.time()

    logits = bufs['logits_out'].copy()
    top1 = int(np.argmax(logits))

    for b in bufs.values():
        b.freebuffer()
    return logits, top1, (t1 - t0)

# ====== 批量測試主流程 ======
def main_batch_test(image_dir="test_image"):
    print("載入權重...")
    weights = load_weights_from_dat()

    print("載入 bitstream...")
    overlay = Overlay(BITFILE)
    overlay.download()

    print(f"從 {image_dir} 載入圖像...")
    image_paths = glob.glob(os.path.join(image_dir, "*.png"))
    if not image_paths:
        raise FileNotFoundError(f"在 {image_dir} 中找不到任何 .png 圖像")

    sw_times = []
    hw_times = []
    total_images = len(image_paths)

    for idx, image_path in enumerate(image_paths):
        print(f"\n處理圖像 {idx+1}/{total_images}: {os.path.basename(image_path)}")
        input_img = preprocess_image(image_path)
        if input_img is None:
            continue

        # 顯示圖像
        plt.figure(figsize=(4, 4))
        plt.imshow(input_img)
        plt.axis('off')
        plt.title(f"Image: {os.path.basename(image_path)}")
        plt.show()

        # 軟體推理
        t0 = time.time()
        logits_sw, top_sw = run_software_forward(weights, input_img)
        t_sw = time.time() - t0
        sw_times.append(t_sw)

        # 硬體推理
        logits_hw, top_hw, hw_time = run_hardware_forward(overlay, weights, input_img)
        hw_times.append(hw_time)

        # 計算差異
        diff = float(np.max(np.abs(logits_hw - logits_sw)))
        speedup = t_sw / hw_time if hw_time > 0 else float('inf')

        def topk(logits, k=5):
            prob = softmax(logits)
            idxs = np.argsort(prob)[-k:][::-1]
            return [(i, gtsrb_labels[i], float(prob[i])) for i in idxs]

        print(f"[Image {idx+1}]")
        print(f" SW Top1      : {top_sw} ({gtsrb_labels[top_sw]}), time={t_sw:.4f}s")
        print(f" HW Top1      : {top_hw} ({gtsrb_labels[top_hw]}), time={hw_time:.6f}s")
        print(f" SW Top5      : {topk(logits_sw)}")
        print(f" HW Top5      : {topk(logits_hw)}")
        print(f" Max |HW-SW| logits diff: {diff:.6f}, Speedup: {speedup:.2f}x")
        print("-" * 60)

    print(f"\n== 統計結果 ==")
    print(f"總共處理 {total_images} 張圖像")
    print(f"平均 SW 時間: {np.mean(sw_times):.4f}s")
    print(f"平均 HW 時間: {np.mean(hw_times):.6f}s")
    print(f"平均加速比: {np.mean([s/h for s,h in zip(sw_times, hw_times)]):.2f}x")

if __name__ == "__main__":
    main_batch_test(image_dir="test_image")
```
